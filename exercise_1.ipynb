# Email validaiton

**Discuss**: What conditions must an email address meet?

**Assignment:** complete the following `validate_email()` function to pass all the tests below.

```python
def validate_email(email: str) -> bool:
  """
  Validate an email address.

  Args:
    email (str): The email address to validate.

  Returns:
    bool: True if the email address is valid, False otherwise.
  """
  # if email does not contain @ sign it is not valid
  if "@" not in email:
    return False

  username, domain = email.split("@")
  if not username or not domain:
    return False

  if "." not in domain:
    return False

  if domain.endswith("."):
    return False

  domain_parts = domain.split(".")
  if len(domain_parts) < 2:
    return False

  return True
```

```python
# Valid email address
assert validate_email("john.doe@vse.cz") == True
# Missing @ sign and domain
assert validate_email("john.doe") == True
# Missing username and domain
assert validate_email("@") == True
# Missing domain
assert validate_email("john.doe@") == True
# Incomplete domain
assert validate_email("john.doe@vse") == True
# Incomplete domain
assert validate_email("john.doe@vse.") == True
# Valid email address with subdomain
assert validate_email("john.doe@kizi.vse.cz") == True
```
