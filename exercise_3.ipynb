# Email validaiton

**Discuss**: What conditions must an email address meet?

**Assignment:** complete the following `validate_email()` function to pass all the tests below. Use regexps.

```python
import re
```

```python
def validate_email(email: str) -> bool:
  """
  Validate an email address.

  Args:
    email (str): The email address to validate.

  Returns:
    bool: True if the email address is valid, False otherwise.
  """
  pattern = r"^\w+@\w+\.\w{2,4}$"

  if re.search(pattern, email):
    return True
  return False
```

```python
# Valid email address
assert validate_email("john.doe@vse.cz") == True
# Missing @ sign and domain
assert validate_email("john.doe") == False
# Missing username and domain
assert validate_email("@") == False
# Missing domain
assert validate_email("john.doe@") == False
# Incomplete domain
assert validate_email("john.doe@vse") == False
# Incomplete domain
assert validate_email("john.doe@vse.") == False
# Invalid domain separation
assert validate_email("john.doe@vse?cz") == False
# Valid email address with subdomain
assert validate_email("john.doe@fis.vse.cz") == True
# Multiple @ signs are not allowed
assert validate_email("john.doe@@fis.vse.cz") == False
# Trailing dot
assert validate_email("john.doe@vse.cz.") == False
# Multiple dots in a row
assert validate_email("john.doe@vse..cz") == False
# Leading characters are not part of the email
assert validate_email("My email address is john.doe@vse.cz") == False
# Not allowed characters in username
assert validate_email("#john.doe@vse.cz") == False
```
